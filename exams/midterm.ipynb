## Question 4: Berkeley 5k

````python
proportions = [2, 4, 4, 14, 12, 16, 12, 8, 10, 8, 1, 3, 3, 2, 1]
bins = np.arange(20, 20 + 2 * len(proportions) + 1, 2)
data = []
for p, b in zip(proportions, bins):
    bounds = np.array([b] * p * 2)
    noise = np.round(np.random.random(p * 2), 2)
    data.extend(bounds + noise)
t = Table().with_columns('Times', data)
wide = make_array(20, 22, 26, 28, 30, 32, 34, 36, 38, 40, 50)
t.hist('Times', bins=wide, unit='minute')
x_axis = plt.xticks(np.arange(20, 51, 2))
````
